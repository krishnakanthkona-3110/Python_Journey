```python
l1 = [10,0,'hello', None]

for e in l1:
  try:
    print(f'current element - {e}')
    result = 5/int(e)
    print(f'result - {result}')

  except Exception as exp:
    print(f'Exception is - {exp}')

    print('*'*50)
```

```python
l1 = [10,0,'hello', None]

for e in l1:
  try:
    print(f'current element - {e}')
    result = 5/int(e)
    print(f'result - {result}')

  except ZeroDivisionError as z:
    print(f'Adding 1 to the list element and now again dividing: 5')

  except Exception as exp:
    print(f'Exception is - {exp}')

  print('*'*50)
```
