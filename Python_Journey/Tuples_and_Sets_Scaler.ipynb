```python
# check if a string is binary or not
```

```python
def check_binary(a):
  for i in a:
    if not (i=='0' or i=='1'):
      return 'Not a binary string'
  return 'Binary string'
```

```python
print(check_binary('1010'))
print(check_binary('101012'))
```

```python
# Using set
def check(b):
  s= set(b)
  if s=={'0','1'}:
    return 'Binary string'
  else:
    return 'Not a binary string'
```

```python
check('01111')
```
